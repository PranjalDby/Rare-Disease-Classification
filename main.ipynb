```python
import os
from remotezip import RemoteZip

image_path = os.path.abspath(os.curdir) + "/images"
print(image_path)
IMAGE_DIR = image_path
os.makedirs(IMAGE_DIR, exist_ok=True)

# 1. Collect missing filenames
split_dfs = [
    training_ds,
    validation_ds,
    testing_ds
]

missing = set()
for split_df in split_dfs:
    for filename in split_df["Image Index"].astype(str):
        if not os.path.isfile(os.path.join(IMAGE_DIR, filename)):
            missing.add(filename)

print("Unique images to recover:", len(missing))


# 2. Scan and extract
for i in range(1, 13):
    if not missing:
        break

    archive_name = f"images_{i:03d}.zip"
    url = (
        "https://huggingface.co/datasets/"
        "alkzar90/NIH-Chest-X-ray-dataset/"
        f"resolve/main/data/images/{archive_name}"
    )

    print(f"\nChecking {archive_name}...")
    print(f"Target URL: {url}") # This will print out exactly what Python sees

    try:
        with RemoteZip(url) as z:
            members = {
                os.path.basename(name): name
                for name in z.namelist()
                if name.lower().endswith(".png")
            }

            matches = missing.intersection(members.keys())
            print(f"Matching missing images to download: {len(matches)}")

            for filename in matches:
                image_bytes = z.read(members[filename])

                output_path = os.path.join(IMAGE_DIR, filename)
                with open(output_path, "wb") as f:
                    f.write(image_bytes)

                missing.remove(filename)

        print("Images still missing:", len(missing))

    except Exception as e:
        print(f"Could not scan {archive_name}: {e}")

print("\nRecovery scan complete.")
print("Final images still missing:", len(missing))
```

```python
IMG_SIZE = (224, 224)
BATCH_SIZE = 16
AUTOTUNE = tf.data.AUTOTUNE
def image_preprocessing(img_path,label):
     # Read image
    image = tf.io.read_file(img_path)

    # Decode PNG
    image = tf.image.decode_png(image, channels=3)

    # Resize
    image = tf.image.resize(image, IMG_SIZE)

    # Convert uint8 → float32 and normalize
    image = tf.cast(image, tf.float32) / 255.0

    return image, label


#creating the tensorflow dataset
def make_dataset(split_df, training=False):
    paths = split_df["image_path"].values
    labels = split_df["label"].astype("int32").values

    ds = tf.data.Dataset.from_tensor_slices((paths, labels))
    ds = ds.map(image_preprocessing, num_parallel_calls=AUTOTUNE)

    if training:
        ds = ds.shuffle(
            buffer_size=len(split_df),
            reshuffle_each_iteration=True
        )

    ds = ds.batch(BATCH_SIZE)
    ds = ds.prefetch(AUTOTUNE)

    return ds

# Add the missing label column to the test split
if "label" not in testing_ds.columns:
    testing_ds["label"] = (
        testing_ds["Finding Labels"].map(class_to_id)
    )

if "label" not in training_ds.columns:
  training_ds['label'] = training_ds['Finding Labels'].map(class_to_id)

if "label" not in validation_ds.columns:
  validation_ds['label'] = validation_ds['Finding Labels'].map(class_to_id)


# Validate labels
assert testing_ds["label"].notna().all(), \
    "Some test labels could not be mapped!"

testing_ds["label"] = (
    testing_ds["label"].astype("int32")
)

# Rebuild datasets
train_ds = make_dataset(training_ds, training=True)
val_ds = make_dataset(validation_ds)
test_ds = make_dataset(testing_ds)

# # Verify
# images, labels = next(iter(train_ds))
```

```python
#building our episodic pipeline which provides one class support and query images
NUMCLASSES = 5
IMG_WIDTH = 224;IMG_HEIGHT = 224

#our image paths
class_paths = []

for class_id in range(NUMCLASSES):
    
    paths = training_ds.loc[
        training_ds["label"] == class_id,
        "image_path"
    ].values
    
    class_paths.append(paths.astype(str))
    
    print(
        class_id,
        class_names[class_id],
        len(paths)
    )
    
#creating the tensor for paths per classes
class_path_tensors = [
    tf.constant(paths, dtype=tf.string)
    for paths in class_paths
]

### loading the image

def load_image(path):
    image = tf.io.read_file(path)
    image = tf.io.decode_png(image,channels=3)
    image = tf.image.resize(
        image,
        [IMG_HEIGHT, IMG_WIDTH]
    )

    image = tf.cast(
        image,
        tf.float32
    ) / 255.0
    
    
    return image



##### episode generator each episode of size N_way * (N_shot + N_query)
N_WAY = 5 #no of classes
N_SHOT = 5
N_QUERY = 5

EPISODE_SIZE = N_WAY * (N_SHOT + N_QUERY)

def episode_generator(tensors, n_way, n_shot, n_query):

    while True:

        support_images = []
        support_labels = []

        query_images = []
        query_labels = []

        for class_id in range(n_way):

            paths = class_path_tensors[class_id]

            total_needed = n_shot + n_query

            # Randomly select paths
            shuffled_paths = tf.random.shuffle(paths)

            selected_paths = tf.slice(
                shuffled_paths,
                [0],
                [total_needed]
            )

            # Support paths
            support_paths = tf.slice(
                selected_paths,
                [0],
                [n_shot]
            )

            # Query paths
            query_paths = tf.slice(
                selected_paths,
                [n_shot],
                [n_query]
            )

            # Load support images
            for path in support_paths:
                support_images.append(load_image(path))
                support_labels.append(class_id)

            # Load query images
            for path in query_paths:
                query_images.append(load_image(path))
                query_labels.append(class_id)

        yield (
            tf.stack(support_images),
            tf.constant(support_labels, dtype=tf.int32),
            tf.stack(query_images),
            tf.constant(query_labels, dtype=tf.int32)
        )
```

```python
# ============================================================
# COMPLETE PROTONET + GRADIO CHEST X-RAY CLASSIFIER
# ============================================================

import tensorflow as tf
import numpy as np
import gradio as gr

from collections import Counter

from tensorflow.keras.applications.resnet50 import preprocess_input


# ============================================================
# 1. CONFIGURATION
# ============================================================

MODEL_PATH = (
    "/workspace/DeepLearningProject/"
    "saved_models/protonet_model.keras"
)

N_WAY = 5
N_SHOT = 5
N_QUERY = 5

NUM_EPISODES = 20

IMAGE_SIZE = (224, 224)

class_names = [
    "Hernia",
    "Pneumonia",
    "Fibrosis",
    "Edema",
    "Emphysema"
]

class_to_id = {
    name: i
    for i, name in enumerate(class_names)
}


# ============================================================
# 2. LOAD PROTONET MODEL
# ============================================================

print("=" * 60)
print("Loading ProtoNet model...")
print("=" * 60)

protonet_model = tf.keras.models.load_model(
    MODEL_PATH,
    compile=False,
    safe_mode=False
)

print("ProtoNet loaded successfully.")

print(
    "Model input:",
    protonet_model.input_shape
)

print(
    "Model output:",
    protonet_model.output_shape
)


# ============================================================
# 3. CREATE EMBEDDING MODEL
# ============================================================

# Your model structure is:
#
# Input
#   ↓
# Augmentation
#   ↓
# ResNet50
#   ↓
# GlobalAveragePooling2D
#   ↓
# Dense(126)       <-- EMBEDDING
#   ↓
# Dense(5)         <-- CLASSIFIER
#
# Therefore layers[-2] is Dense(126).

embedding_model = tf.keras.Model(
    inputs=protonet_model.input,
    outputs=protonet_model.layers[-2].output
)

print(
    "Embedding output:",
    embedding_model.output_shape
)

# Expected:
# (None, 126)


# ============================================================
# 4. LOAD IMAGE
# ============================================================

def load_image(path):

    image = tf.io.read_file(path)

    image = tf.image.decode_png(
        image,
        channels=3
    )

    image = tf.image.resize(
        image,
        IMAGE_SIZE
    )

    image = tf.cast(
        image,
        tf.float32
    )

    image = image / 255.0

    return image


# ============================================================
# 5. CREATE TEST CLASS PATH TENSORS
# ============================================================

# If you ALREADY have:
#
# test_class_path_tensors
#
# from your notebook, this section will simply use it.
#
# Otherwise, create it from your test_preprocessed_ds.

try:

    class_path_tensors_test

    print(
        "Using existing test_class_path_tensors."
    )

except NameError:

    print(
        "test_class_path_tensors not found."
    )

    print(
        "Creating it from test_preprocessed_ds..."
    )

    class_path_tensors_test = []

    for class_name in class_names:

        class_df = testing_ds[
            testing_ds["label"]
            == class_name
        ]

        paths = class_df[
            "image_path"
        ].values

        class_path_tensors_test.append(
            tf.constant(
                paths,
                dtype=tf.string
            )
        )

    print(
        "Created class_path_tensors_test."
    )


# ============================================================
# 6. EPISODE GENERATOR
# ============================================================

def episode_generator(
    class_path_tensors,
    n_way,
    n_shot,
    n_query
):

    while True:

        support_images = []
        support_labels = []

        query_images = []
        query_labels = []

        # --------------------------------------------
        # Select samples from every class
        # --------------------------------------------

        for class_id in range(n_way):

            paths = class_path_tensors[
                class_id
            ]

            # Randomize paths
            shuffled_paths = tf.random.shuffle(
                paths
            )

            # Need n_shot + n_query
            selected_paths = shuffled_paths[
                :n_shot + n_query
            ]

            support_paths = selected_paths[
                :n_shot
            ]

            query_paths = selected_paths[
                n_shot:n_shot + n_query
            ]

            # ----------------------------------------
            # Support images
            # ----------------------------------------

            for path in support_paths:

                support_images.append(
                    load_image(path)
                )

                support_labels.append(
                    class_id
                )

            # ----------------------------------------
            # Query images
            # ----------------------------------------

            for path in query_paths:

                query_images.append(
                    load_image(path)
                )

                query_labels.append(
                    class_id
                )

        # --------------------------------------------
        # Return complete episode
        # --------------------------------------------

        yield (
            tf.stack(support_images),
            tf.constant(
                support_labels,
                dtype=tf.int32
            ),
            tf.stack(query_images),
            tf.constant(
                query_labels,
                dtype=tf.int32
            )
        )


# ============================================================
# 7. CREATE PROTOTYPES
# ============================================================

def create_prototypes(
    support_images,
    support_labels
):

    # --------------------------------------------
    # Get support embeddings
    # --------------------------------------------

    support_embeddings = embedding_model(
        support_images,
        training=False
    )

    # --------------------------------------------
    # Calculate one prototype per class
    # --------------------------------------------

    prototypes = []

    for class_id in range(N_WAY):

        class_mask = (
            support_labels == class_id
        )

        class_embeddings = tf.boolean_mask(
            support_embeddings,
            class_mask
        )

        prototype = tf.reduce_mean(
            class_embeddings,
            axis=0
        )

        prototypes.append(
            prototype
        )

    prototypes = tf.stack(
        prototypes
    )

    return prototypes


# ============================================================
# 8. PREPROCESS GRADIO IMAGE
# ============================================================

def preprocess_query(image):

    # Gradio gives numpy array
    # Convert to float32

    image = tf.cast(
        image,
        tf.float32
    )

    # Handle RGBA
    if image.shape[-1] == 4:

        image = image[:, :, :3]

    # Resize
    image = tf.image.resize(
        image,
        IMAGE_SIZE
    )

    # IMPORTANT:
    #
    # Your saved model contains:
    #
    # preprocess_input(img * 255.0)
    #
    # internally.
    #
    # Therefore we pass [0,1] here.

    image = image / 255.0

    # Add batch dimension
    image = tf.expand_dims(
        image,
        axis=0
    )

    return image


# ============================================================
# 9. PREDICT ONE EPISODE
# ============================================================

def predict_one_episode(
    query_image
):

    # --------------------------------------------
    # Generate random support set
    # --------------------------------------------

    episode_gen = episode_generator(
        class_path_tensors_test,
        N_WAY,
        N_SHOT,
        N_QUERY
    )

    (
        support_images,
        support_labels,
        _,
        _
    ) = next(
        episode_gen
    )

    # --------------------------------------------
    # Create prototypes
    # --------------------------------------------

    prototypes = create_prototypes(
        support_images,
        support_labels
    )

    # --------------------------------------------
    # Query embedding
    # --------------------------------------------

    query_embedding = embedding_model(
        query_image,
        training=False
    )

    # --------------------------------------------
    # Euclidean squared distance
    # --------------------------------------------

    distances = tf.reduce_sum(
        tf.square(
            query_embedding[
                :,
                tf.newaxis,
                :
            ]
            -
            prototypes[
                tf.newaxis,
                :,
                :
            ]
        ),
        axis=-1
    )

    # Convert:
    #
    # (1, 5)
    #
    # to:
    #
    # (5,)

    distances = distances.numpy()[0]

    # --------------------------------------------
    # Select closest prototype
    # --------------------------------------------

    predicted_id = int(
        np.argmin(distances)
    )

    return (
        predicted_id,
        distances
    )


# ============================================================
# 10. RUN STABILITY TEST
# ============================================================

def run_stability_test(
    query_image
):

    predictions = []

    all_distances = []

    print()
    print("=" * 60)
    print("RUNNING PROTONET STABILITY TEST")
    print("=" * 60)

    for episode_number in range(
        NUM_EPISODES
    ):

        predicted_id, distances = (
            predict_one_episode(
                query_image
            )
        )

        predictions.append(
            predicted_id
        )

        all_distances.append(
            distances
        )

        print(
            f"Episode "
            f"{episode_number + 1:02d}: "
            f"{class_names[predicted_id]:<12} "
            f"| Min distance = "
            f"{np.min(distances):.4f}"
        )

    print("=" * 60)

    return (
        predictions,
        np.array(
            all_distances
        )
    )


# ============================================================
# 11. GRADIO PREDICTION FUNCTION
# ============================================================

def gradio_predict(image):

    # --------------------------------------------
    # Check image
    # --------------------------------------------

    if image is None:

        return (
            """
# ⚠️ No Image

Please upload a chest X-ray image.
""",
            [],
            "Waiting for image..."
        )

    # --------------------------------------------
    # Preprocess
    # --------------------------------------------

    query_image = preprocess_query(
        image
    )

    print()
    print(
        "Query shape:",
        query_image.shape
    )

    print(
        "Query dtype:",
        query_image.dtype
    )

    print(
        "Query range:",
        float(
            tf.reduce_min(
                query_image
            ).numpy()
        ),
        "→",
        float(
            tf.reduce_max(
                query_image
            ).numpy()
        )
    )

    # --------------------------------------------
    # Check embedding
    # --------------------------------------------

    query_embedding = embedding_model(
        query_image,
        training=False
    )

    print(
        "Embedding shape:",
        query_embedding.shape
    )

    # --------------------------------------------
    # Run 20 episodes
    # --------------------------------------------

    (
        predictions,
        distance_array
    ) = run_stability_test(
        query_image
    )

    # --------------------------------------------
    # Count predictions
    # --------------------------------------------

    prediction_counts = Counter(
        predictions
    )

    # Most common
    final_id, final_count = (
        prediction_counts.most_common(1)[0]
    )

    final_class = class_names[
        final_id
    ]

    # --------------------------------------------
    # Stability
    # --------------------------------------------

    stability = (
        final_count /
        NUM_EPISODES
    ) * 100

    # --------------------------------------------
    # Mean distances
    # --------------------------------------------

    mean_distances = np.mean(
        distance_array,
        axis=0
    )

    # --------------------------------------------
    # Create table
    # --------------------------------------------

    rows = []

    for class_id, class_name in enumerate(
        class_names
    ):

        count = prediction_counts.get(
            class_id,
            0
        )

        frequency = (
            count /
            NUM_EPISODES
        ) * 100

        mean_distance = (
            mean_distances[class_id]
        )

        rows.append({
            "class": class_name,
            "count": count,
            "frequency": frequency,
            "distance": mean_distance
        })

    # --------------------------------------------
    # Sort by prediction frequency
    # --------------------------------------------

    rows.sort(
        key=lambda x: (
            -x["frequency"],
            x["distance"]
        )
    )

    # --------------------------------------------
    # Gradio table
    # --------------------------------------------

    table = []

    for row in rows:

        table.append([
            row["class"],
            f'{row["count"]}/{NUM_EPISODES}',
            f'{row["frequency"]:.1f}%',
            f'{row["distance"]:.2f}'
        ])

    # --------------------------------------------
    # Find closest class by distance
    # --------------------------------------------

    closest_class_id = int(
        np.argmin(
            mean_distances
        )
    )

    closest_class = class_names[
        closest_class_id
    ]

    closest_distance = (
        mean_distances[
            closest_class_id
        ]
    )

    # ========================================================
    # RESULT MARKDOWN
    # ========================================================

    result = f"""
# 🩻 ProtoNet Prediction

# **{final_class}**

## 🎯 Stability: **{stability:.0f}%**

The model predicted **{final_class}** in

### **{final_count} / {NUM_EPISODES} episodes**

---

## 📊 Prediction Frequency

"""

    # Add prediction frequency
    for row in rows:

        if row["count"] > 0:

            result += (
                f"**{row['class']}** — "
                f"{row['count']}/{NUM_EPISODES} "
                f"({row['frequency']:.1f}%)\n\n"
            )

    result += f"""
---

## 📐 Prototype Distance

The smallest mean prototype distance was:

### **{closest_class} — {closest_distance:.2f}**

Lower distance means the query embedding is closer to
that class prototype in the learned embedding space.

---

### ⚠️ Important

The stability percentages are **prediction frequencies**
across randomly sampled support episodes.

They are **not clinical probabilities**.

This model is intended for research and demonstration,
not clinical diagnosis.
"""

    # --------------------------------------------
    # Status
    # --------------------------------------------

    status = (
        f"✅ Final prediction: **{final_class}** "
        f"| Stability: **{stability:.0f}%**"
    )

    return (
        result,
        table,
        status
    )


# ============================================================
# 12. GRADIO INTERFACE
# ============================================================

with gr.Blocks(
    title="ProtoNet Chest X-Ray Classifier"
) as demo:

    # ========================================================
    # HEADER
    # ========================================================

    gr.Markdown(
        """
# 🩻 ProtoNet Chest X-Ray Classifier

### Few-Shot Disease Classification using Prototypical Networks

Upload a chest X-ray image. The model evaluates the image
using **20 randomly sampled 5-way 5-shot episodes**.

The final prediction is the class that appears most
frequently across the episodes.
"""
    )

    gr.Markdown("---")

    # ========================================================
    # INPUT / OUTPUT
    # ========================================================

    with gr.Row():

        # ----------------------------------------------------
        # LEFT
        # ----------------------------------------------------

        with gr.Column(
            scale=1
        ):

            image_input = gr.Image(
                type="numpy",
                image_mode="RGB",
                label="Upload Chest X-Ray",
                height=400
            )

            analyze_button = gr.Button(
                "🔍 Analyze X-Ray",
                variant="primary",
                size="lg"
            )

        # ----------------------------------------------------
        # RIGHT
        # ----------------------------------------------------

        with gr.Column(
            scale=1
        ):

            prediction_output = gr.Markdown(
                """
# Prediction

Upload an X-ray and click:

## 🔍 Analyze X-Ray
"""
            )

    gr.Markdown("---")

    # ========================================================
    # STABILITY RESULTS
    # ========================================================

    gr.Markdown(
        """
## 📊 Episode-Level Stability

Each episode uses a new randomly sampled **5-shot support
set** to construct the class prototypes.
"""
    )

    stability_output = gr.Dataframe(
        headers=[
            "Class",
            "Predictions",
            "Frequency",
            "Mean Prototype Distance"
        ],
        datatype=[
            "str",
            "str",
            "str",
            "str"
        ],
        interactive=False,
        label="20-Episode Evaluation"
    )

    # ========================================================
    # STATUS
    # ========================================================

    status_output = gr.Markdown(
        "🟢 Ready"
    )

    # ========================================================
    # BUTTON EVENT
    # ========================================================

    analyze_button.click(
        fn=gradio_predict,
        inputs=image_input,
        outputs=[
            prediction_output,
            stability_output,
            status_output
        ]
    )


# ============================================================
# 13. LAUNCH
# ============================================================

print()
print("=" * 60)
print("STARTING GRADIO")
print("=" * 60)
print("Port: 6007")
print("=" * 60)

demo.launch(
    server_name="0.0.0.0",
    share=True
)
```
